```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1,2,3])
y = np.array([1,2,0])

#Esta funcion ordena la matriz para tener x²+x+1
A = np.vstack([x**2, x, np.ones(len(x))]).T

#Esta funcion calcula los coeficientes
coef = np.linalg.solve(A, y)

a, b, c = coef

xx = np.array([1,2,3])
yy = np.array([1,2,0])
x = np.linspace(0,4,100)
f = lambda t: a*t**2+b*t+c
plt.plot(xx,yy,'o')
plt.plot(x,f(x))
plt.show()
```
