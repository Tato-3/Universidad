```python
#Ejercicio 1

print(0.3 + 0.25)
print(0.3-0.25)
```

No se obtienen los resultados esperados por ambas partes, en el caso de la suma tenemos lo que esperabamos. Por el otro lado, en el caso de la resta nos devuelve una aproximacion al numero 0.05.

La expresion de 0.25 en terminos de su mantisa y exponente es:
Mantisa: 1, Exponente: -2

La expresion de 0.3 en terminos de su mantisa y exponente es:
Mantisa: 0.2 aproximado, Exponente: -2
Lo que ocurre en este caso, es que la mantisa de 0.3 es infinita.

```python
#Ejercicio 2
import numpy as np
import matplotlib.pyplot as plt

print(np.sqrt(2)**2-2)

def funcA(x):
    return (np.sqrt((2*x**2)+1))-1

def funcB(x):
    return (2*x**2) / ((np.sqrt((2*x**2)+1))+1)

x = np.linspace(10**14, 10**16, 100)

y_a = funcA(x)
y_b = funcB(x)

plt.plot(x, y_a)
plt.title("Funcion a")
plt.xlabel("x")
plt.ylabel("y")
plt.show()

plt.plot(x, y_b)
plt.title("Funcion b")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
#Ejercicio 5

def matricesIguales(A,B):
    if A.shape != B.shape: return False
    return np.allclose(A,B)

A = np.array([[4.,2.,1.], [2.,7.,9.], [0.,5.,22/3]])
L = np.array([[1.,0.,0.], [0.5,1.,0.], [0.,5/6,1.]])
U = np.array([[4.,2.,1.], [0.,6.,8.5], [0.,0.,0.25]])

print(matricesIguales(A,L@U))
```

```python
#Ejercicio 6

def esSimetrica(A):
    res = True
    if not(A.shape[0] == A.shape[1]): return False
    AT = A.T
    n, m = A.shape
    for i in range(n):
        for j in range(m):
            if A[i,j] != AT[i,j]: 
                res = False
                break
    return res

A = np.array(np.random.rand(4,4))

print(esSimetrica(A.T@A))
print(esSimetrica(A.T@((A*0.25)/0.25)))
print(esSimetrica(A.T@((A*0.2)/0.2)))
```
